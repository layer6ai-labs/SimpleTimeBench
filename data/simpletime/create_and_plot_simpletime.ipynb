# Generate and Plot simpletime Data

This notebook demonstrates how to generate synthetic time series datasets using `simpletime` and then load and visualize them.

```python
import os
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from datasets import load_from_disk

# Add current directory to path to import local modules
current_dir = os.getcwd()
if current_dir not in sys.path:
    sys.path.append(current_dir)

# Define dataset storage path if not already set
if 'SYNTHETIC' not in os.environ:
    os.environ['SYNTHETIC'] = str(Path(current_dir) / 'datasets')
    print(f"Setting SYNTHETIC to {os.environ['SYNTHETIC']}")
else:
    print(f"Using existing SYNTHETIC path: {os.environ['SYNTHETIC']}")

import simpletime
```

## 1. Generate Datasets
We define the arguments/configuration for the generation and call `simpletime.main`.

```python
class Args:
    length = 256
    num_samples = 1024
    num_series = 2
    lag_amount = 48
    skip_load_check = True
    verbose = False
    nprint = 3
    plot_examples = False # We will plot separately in this notebook
    nplt = 3
    # Generating a subset of distributions for demonstration
    distributions = [
        #Deterministic Trends
        'constant', 'linear', 'staircase', 'power_law', 'exponential', 
        #Harmonic & Periodic
        'sine', 'sawtooth', 'square', 'triangle', 'noise_patch', 'fourier',
        #Transient Dynamics
        'damped_sine', 'growing_sine', 'chirp',
        # Autocorrelated Processes 
        'ar', 'ma', 'arma', 'sarima', 'garch', 'pink_noise', 'fractional_brownian', 'markov_chain', 
        #I.I.D. White Noise'
        'normal', 'uniform', 'poisson', 'binary', 'student_t', 'lognormal', 'laplace', 'cauchy', 'skew_normal',
        #Stochastic Drift/Shocks
        'random_walk', 'piecewise_constant', 'gbm', 'intermittent', 'impulse', 'logistic', 
    ]
    storage_env_var = 'SYNTHETIC'

args = Args()

print("Starting dataset generation...")
simpletime.main(args)
print("Generation complete.")
```

## 2. Load and Plot Datasets
Now we load the generated Hugging Face datasets from the disk and plot a few examples.

```python
def load_and_plot_individual(dataset_path, distribution_name, num_examples=3):
    try:
        if not os.path.exists(dataset_path):
            print(f"Dataset not found at {dataset_path}")
            return

        dataset = load_from_disk(dataset_path)
        print(f"Successfully loaded {distribution_name} from {dataset_path}")
        
        # The dataset contains 'target' which is a list of lists (variables x time)
        
        fig, axes = plt.subplots(num_examples, 1, figsize=(10, 4 * num_examples), sharex=True)
        if num_examples == 1:
            axes = [axes]
            
        for i in range(num_examples):
            if i >= len(dataset):
                break
            entry = dataset[i]
            target = entry['target']
            item_id = entry.get('item_id', f'Sample {i}')
            
            # Check if univariate (1D list/array) or multivariate (2D)
            # Based on save_hf_dataset, target is [s[col].tolist() ...]. 
            # If 1 var, it might be a single list if the saving logic flattened it. 
            # simpletime.py: d['target'] = d['target'][0] if len==1
            
            ax = axes[i]
            
            if isinstance(target[0], list) or (isinstance(target, np.ndarray) and target.ndim > 1):
                 # Multivariate: List of lists
                 for v_idx, series in enumerate(target):
                     ax.plot(series, label=f'V_{v_idx}')
                 ax.legend()
            else:
                # Univariate
                ax.plot(target, label='Target')
                
            ax.set_title(f"{distribution_name} - {item_id}")
            ax.grid(True)
            
        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"Failed to load or plot {distribution_name}: {e}")


def load_and_plot_all(dataset_path, distribution_name, ax=None, example=0, normalize=True):
    try:
        if ax is None:
            fig, ax = plt.subplots(1, 1,figsize=(10, 4))

        if not os.path.exists(dataset_path):
            print(f"Dataset not found at {dataset_path}")
            return

        dataset = load_from_disk(dataset_path)
        
        # The dataset contains 'target' which is a list of lists (variables x time)
        
        entry = dataset[example]
        target = entry['target']
        item_id = entry.get('item_id', f'Sample {example}')
        
        # Check if univariate (1D list/array) or multivariate (2D)
        # Based on save_hf_dataset, target is [s[col].tolist() ...]. 
        # If 1 var, it might be a single list if the saving logic flattened it. 
        # simpletime.py: d['target'] = d['target'][0] if len==1
                
        if isinstance(target[0], list) or (isinstance(target, np.ndarray) and target.ndim > 1):
                # Multivariate: List of lists
                for v_idx, series in enumerate(target):
                    if normalize:
                        series = np.array(series)
                        series = (series - series.min()) / (series.max() - series.min())
                    ax.plot(series, label=f'{distribution_name}: V_{v_idx}')
        else:
            # Univariate
            if normalize:
                target = np.array(target)
                target = (target - target.min()) / (target.max() - target.min())
            ax.plot(target, label=f'{distribution_name}')
            
    except Exception as e:
        print(f"Failed to load or plot {distribution_name}: {e}")


def normalize_list(data):
    data = np.array(data)
    return (data - data.min() + 1e-8) / (data.max() - data.min() + 1e-8)
    
def load_and_plot_panels(dataset_path, distribution_name, ax, example=0, num_examples=3, normalize=True):
    try:
        if not os.path.exists(dataset_path):
            print(f"Dataset not found at {dataset_path}")
            return

        dataset = load_from_disk(dataset_path)
        
        # The dataset contains 'target' which is a list of lists (variables x time)

        for examplei in range(example, example+num_examples):
            entry = dataset[examplei]
            target = entry['target']
            item_id = entry.get('item_id', f'Sample {examplei}')
            
            # Check if univariate (1D list/array) or multivariate (2D)
            # Based on save_hf_dataset, target is [s[col].tolist() ...]. 
            # If 1 var, it might be a single list if the saving logic flattened it. 
            # simpletime.py: d['target'] = d['target'][0] if len==1
                    
            if isinstance(target[0], list) or (isinstance(target, np.ndarray) and target.ndim > 1):
                    # Multivariate: List of lists
                    for v_idx, series in enumerate(target):
                        if normalize:
                            series = normalize_list(series)
                        ax.plot(series, label=f'V_{v_idx}')
                    ax.legend()
    
            else:
                # Univariate
                if normalize:
                    target = normalize_list(target)
                ax.plot(target, label=f'{distribution_name}')
    
        ax.set_title(distribution_name)
        ax.grid(True)

    except Exception as e:
        print(f"Failed to load or plot {distribution_name}: {e}")
```

```python
import math
storage_path = Path(os.environ['SYNTHETIC'])

nx = 2
num_examples = 2
normalize=True
# Iterate over generated datasets
# simpletime saves to: {storage_path}/{dataset_str}/{distribution}
dataset_types = ['univariate']#, 'multivariate', 'multivariate_lagged']

regimes = {
    'Deterministic Trends': ['constant', 'linear', 'exponential', 'staircase'],
    'Harmonic & Periodic': ['sine', 'sawtooth', 'square', 'triangle', 'noise_patch', 'fourier'],
    'Transient Dynamics': ['damped_sine', 'growing_sine', 'chirp'],
    'Autocorrelated Processes': ['ar', 'ma', 'arma', 'sarima', 'garch', 'pink_noise', 'fractional_brownian', 'markov_chain'],
    'I.I.D. White Noise': ['normal', 'uniform', 'poisson', 'binary', 'student_t', 'lognormal', 'laplace', 'cauchy', 'skew_normal'],
    'Stochastic Drift/Shocks': ['random_walk', 'piecewise_constant', 'gbm', 'intermittent', 'impulse', 'logistic']
}

for ds_type in dataset_types:
    type_path = storage_path / ds_type

    if not type_path.exists():
        print(f"Skipping {ds_type} - directory does not exist")
        continue
        
    print(f"\n=== Plotting {ds_type} ===")
    for regime, distributions in regimes.items():

        ny = math.ceil(len(distributions)/nx)
        fig, axes = plt.subplots(ny, nx, figsize=(6*nx, 4*ny), squeeze=False)
        fig.suptitle(f"{regime}, normalized={normalize}", fontsize=16, y=1.005)
        
        for i, dist in enumerate(distributions):
            ix = i % nx
            iy = i // nx
            dist_path = type_path / dist
            # Check if the path is a directory (it should be for load_from_disk)
            if dist_path.exists() and dist_path.is_dir():
                load_and_plot_panels(str(dist_path), dist, example=0, num_examples=num_examples, ax=axes[iy, ix], normalize=normalize)


        plt.tight_layout()
        plt.show()

```

```python
# Gemini generic example, unsure if will use, just brainstorming a "scorecard"

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

def generate_simpleTS_scorecard(results_dict):
    """
    results_dict: dict mapping regime name to a performance score (0-1)
    Example: {'Deterministic': 0.99, 'Harmonic': 0.95, ...}
    """
    # 1. Prepare Data
    df = pd.DataFrame(list(results_dict.items()), columns=['Regime', 'Score'])
    df = df.set_index('Regime')

    # 2. Setup Figure
    fig = plt.figure(figsize=(12, 8), constrained_layout=True)
    gs = fig.add_gridspec(2, 2)
    
    # --- SUBPLOT 1: Heatmap Scorecard ---
    ax1 = fig.add_subplot(gs[0, :])
    sns.heatmap(df.T, annot=True, cmap="RdYlGn", vmin=0, vmax=1, 
                cbar=False, linewidths=2, ax=ax1, annot_kws={"size": 14})
    ax1.set_title("simpletime: Regime Performance (Normalized Accuracy)", fontsize=16, pad=20)
    ax1.set_xlabel("")
    ax1.set_yticklabels([])

    # --- SUBPLOT 2: Diagnostic (Example of "The Human Gap") ---
    ax2 = fig.add_subplot(gs[1, 0])
    t = np.linspace(0, 10, 100)
    truth = 0.5 * t  # Linear Trend
    pred = 0.5 * t + np.sin(t*2) * 0.2 # Model "hallucinating" cycles
    ax2.plot(t, truth, 'k--', label="Ground Truth (Linear)")
    ax2.plot(t, pred, 'r', label="Model Prediction")
    ax2.set_title("Diagnostic: Inductive Bias Failure")
    ax2.legend()

    # --- SUBPLOT 3: Diagnostic (The Noise Floor) ---
    ax3 = fig.add_subplot(gs[1, 1])
    noise = np.random.normal(0, 1, 100)
    ax3.plot(noise, color='gray', alpha=0.5, label="I.I.D. White Noise")
    ax3.axhline(0, color='blue', label="Optimal Prediction (Mean)")
    ax3.set_title("Diagnostic: Noise Floor Control")
    ax3.legend()

    plt.show()

# Example usage:
results = {
    'Deterministic': 0.99,
    'Harmonic': 0.92,
    'Transient': 0.75,
    'Autocorrelated': 0.40,
    'White Noise': 0.05,
    'Stochastic': 0.02
}

generate_simpleTS_scorecard(results)
```
