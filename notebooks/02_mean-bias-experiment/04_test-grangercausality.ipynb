```python
from statsmodels.tsa.stattools import grangercausalitytests
```

```python
import numpy as np

np.random.seed(42)
var1 = np.random.randn(1000)
var2 = np.roll(var1, 3)
var2[:3] = 0  # fill the first three values with zeros due to lag
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(var1, label='var1', alpha=0.6)
plt.plot(var2, label='var2', alpha=0.6)
plt.legend()
plt.title('var1 and var2 Time Series')
plt.xlabel('Index')
plt.ylabel('Value')
plt.show()
```

```python
data = np.column_stack((var1[2:], var2[2:]))
```

```python
grangercausalitytests(data, maxlag=11)
```
