```python
import os
from data.gift_eval.src.gift_eval.data import Dataset
import numpy as np
import pandas as pd
from dotenv import load_dotenv
from pathlib import Path
import warnings
import os
import pickle
import sys
from notebooks.covar_examine.covariate_utils import analyze_variate_contributions, plot_analysis_results
from tqdm import tqdm
folder_path = "notebooks/covar_examine"
```

```python
# Suppress verbose output from statsmodels
warnings.filterwarnings("ignore")

# Load environment variables
load_dotenv()
database_name = "GIFT_EVAL" # Options: "simpletime", "GIFT_EVAL"
storage_env_var = "SYNTHETIC" if database_name=="simpletime" else "GIFT_EVAL"
# Get the GIFT_EVAL path from environment variables
data_path = os.getenv(storage_env_var)

if data_path:
    # Convert to Path object for easier manipulation
    data_path = Path(data_path)

    # Get all subdirectories (dataset names) in the GIFT_EVAL path
    dataset_names = []
    for dataset_dir in data_path.iterdir():
        if dataset_dir.name.startswith("."):
            continue
        if dataset_dir.is_dir():
            freq_dirs = [d for d in dataset_dir.iterdir() if d.is_dir()]
            if freq_dirs:
                for freq_dir in freq_dirs:
                    dataset_names.append(f"{dataset_dir.name}/{freq_dir.name}")
            else:
                dataset_names.append(dataset_dir.name)

    # print("Available datasets in GIFT_EVAL:")
    # for name in sorted(dataset_names):
    #     print(f"- {name}")
else:
    print(
        "GIFT_EVAL path not found in environment variables. Please check your .env file."
    )
```

```python
max_lag = 100

db_compare = pd.DataFrame()
for dataset_name in tqdm(dataset_names):
    for term in ["short", "medium", "long"]:
        dataset = Dataset(name=dataset_name, term=term, to_univariate=False, storage_env_var=storage_env_var)
        try:
            train_data = next(iter(dataset.training_dataset))
        except Exception as e:
            # print(f"Error loading {dataset_name}/{term}: {e}")
            continue

        if train_data['target'].ndim == 1:
            continue
                
        val_data = next(iter(dataset.validation_dataset))
        test_data = next(iter(dataset.test_data))
        
        arr = np.concatenate([
            data['target'] if data['target'].ndim == 2 else data['target'][np.newaxis, :] for data in [train_data, val_data, test_data[1]]
        ], axis=1)
        arr = np.nan_to_num(arr, nan=0)
        
        if arr.shape[1] > 1e4:
            continue
        
        print(f"Dataset: {dataset_name}/{term} of shape {arr.shape} loaded.")

        analysis_results = analyze_variate_contributions(
            ts_data=arr, max_lag=max_lag, verbose=False
        )

        # --- Visualize the Results ---
        if analysis_results:
            plot_analysis_results(analysis_results)

        aggregate_df = pd.DataFrame({
            'dataset': dataset_name, 'term': term, 'num_variates': arr.shape[0], 'length': arr.shape[1],
            **{k: v.mean().mean() for k, v in analysis_results.items()}
        }, index=[0])
        
        db_compare = pd.concat([db_compare, aggregate_df], axis=0)
        db_compare.to_csv(f"{folder_path}/{database_name}_covariate_results.csv")
    
db_compare
```
