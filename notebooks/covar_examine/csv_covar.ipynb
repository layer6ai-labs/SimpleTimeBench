## Analyze Covariate Contribution

```python
Hahn_db_compare = pd.DataFrame()

max_covar_num = 20
max_lag = 30

for folder_path in ["data/synthetic", "data/Hahn_survey"]:
    # Iterate over all files in the folder
    csv_files = glob(f"{folder_path}/*.csv")  # Use glob to find all .csv files
    for file_path in csv_files:
        print(f"Processing file: {file_path}")
        # Load the CSV file into a DataFrame
        csv_data = pd.read_csv(file_path).dropna(axis=1, how='all')
        display(csv_data.head(4))
        
        if len(csv_data) > 1e3:
            print(f"skipping data of shape {csv_data.shape} in {file_path} as it is too large.")
            continue
        
        # Perform analysis on the data
        if not csv_data.empty:
            arr = csv_data.select_dtypes(include=[float, int]).values.T  # Transpose to match expected shape
            arr = np.nan_to_num(arr, nan=0)
            # Drop timestamp-related columns if they exist
            timestamp_columns = ['Date', 'date', 'Time', 'Timestamp']
            csv_data = csv_data.drop(columns=[col for col in timestamp_columns if col in csv_data.columns], errors='ignore')
            covariate_names = csv_data.columns.tolist()
            if arr.shape[0] > max_covar_num:
                arr = arr[:max_covar_num, :]  # Limit to first max_covar_num variates if more exist
                covariate_names = covariate_names[:max_covar_num]  # Adjust names accordingly
                print(f"Limiting analysis to first {max_covar_num} variates in {file_path}.")

            analysis_results = analyze_variate_contributions(
                ts_data=arr, max_lag=max_lag, verbose=False
            )

            # --- Visualize the Results ---
            if analysis_results:
                plot_analysis_results(analysis_results)
                
            aggregate_df = pd.DataFrame({
                'dataset': file_path, **{k: v.mean().mean() for k, v in analysis_results.items()}
            }, index=[0])
            Hahn_db_compare = pd.concat([Hahn_db_compare, aggregate_df], axis=0)
            Hahn_db_compare.to_csv(f"{result_folder_path}/Hahn_covariate_results.csv")

Hahn_db_compare
```
